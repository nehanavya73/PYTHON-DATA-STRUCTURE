2. Explain the difference between mutable and immutable data types with example?
    -# Difference Between Mutable and Immutable Data Types

## Introduction

In programming, data types can be classified as **mutable** or **immutable** based on whether their values can be changed after they are created.

## Mutable Data Types

A **mutable data type** is a data type whose value or contents can be changed after the object is created. Changes can be made without creating a completely new object.

**Examples:** List, Dictionary, and Set in Python.

### Example:

```python
numbers = [10, 20, 30]
numbers[1] = 50

print(numbers)
```

**Output:**

```text
[10, 50, 30]
```

Here, the list is modified after it is created. Therefore, a list is a **mutable** data type.

## Immutable Data Types

An **immutable data type** is a data type whose value cannot be changed after the object is created. If we try to modify it, a new object is created instead.

**Examples:** Integer, Float, String, and Tuple in Python.

### Example:

```python
name = "John"
name = "Mike"

print(name)
```

**Output:**

```text
Mike
```

In this example, the original string `"John"` is not changed. Instead, the variable `name` refers to a new string `"Mike"`. Therefore, strings are **immutable**.

## Difference Between Mutable and Immutable Data Types

| Mutable                             | Immutable           |
| Can be changed after creation.      | Cannot be changed after creation.              |
| Existing object can be modified.    | A new object is created when a change is made. |
| Examples: List, Dictionary, Set.    | Examples: Integer, String, Tuple.              |
| Changes can affect the same object. | Changes do not modify the original object.|

## Conclusion

The main difference between mutable and immutable data types is that **mutable objects can be modified after creation, whereas immutable objects cannot be modified**. Understanding this difference is important because it affects how data is stored, changed, and handled in a program.

14.How does the “in” keyword work differently for lists and dictionaries?
     -# Difference Between `in` for Lists and Dictionaries in Python

The **`in` keyword** in Python is used to check whether a particular element exists in a collection. However, it works differently with lists and dictionaries.

For a **list**, the `in` keyword checks whether a value is present among the elements of the list.

```python
numbers = [10, 20, 30, 40]

print(20 in numbers)
```

The output will be:

```text
True
```

Here, Python searches for `20` in the list and returns `True` because the value is present.

For a **dictionary**, the `in` keyword checks whether a value is present as a **key**, not as a value.

```python
student = {"name": "Rahul", "age": 20}

print("name" in student)
```

The output will be:

```text
True
```

Here, `"name"` is a key in the dictionary, so the result is `True`. If we want to check whether a value exists in a dictionary, we can use the `values()` method.

```python
print("Rahul" in student.values())
```

The output will be:

```text
True
```

In conclusion, when used with a **list**, `in` checks for an element in the list. When used directly with a **dictionary**, `in` checks for a key. This difference is important when working with Python collections.

17.Describe the time complexity of accessing elements in a dictionary?
    -# Time Complexity of Accessing Elements in a Dictionary

In Python, accessing an element from a dictionary using its key generally has a **time complexity of O(1)**, which means **constant time**. This is because dictionaries use a hash table internally. The key is processed using a hash function to determine where the corresponding value is stored.

For example:

```python
student = {"name": "Rahul", "age": 20}

print(student["name"])
```

In this example, Python can quickly find the value associated with the key `"name"` without checking every element in the dictionary.

The **average-case time complexity is O(1)** for accessing, inserting, and deleting dictionary elements. However, in rare situations involving hash collisions, the time complexity can become **O(n)** in the worst case.

Therefore, dictionaries are very efficient when we need to **store and retrieve data using keys**, especially when dealing with a large amount of data.

````python
# 27.Write a code to create a list of 5 random numbers between 1 and 100, sort it in ascending order, and print the result.(replaced).


import random

numbers = [random.randint(1, 100) for _ in range(5)]
numbers.sort()
print(numbers)
````
